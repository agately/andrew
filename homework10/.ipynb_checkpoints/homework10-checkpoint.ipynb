```python
# 2.1.2 Step 3
# The actual values in the DataFrame are all stored as Python strings (type <class 'str'>).
# Even though they look like numbers, pandas reads them as strings because they came from a CSV.
```

```python
# 2.1.3 Step 1

ignore_cols = ["Planet", "Ring System?", "Global Magnetic Field?"]

for col in df.columns:
    if col not in ignore_cols:
        df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
# 2.1.3 Step 2
df.dtypes
```

```python
# 2.1.3 Step 3

# The three different data types now in the DataFrame are float64, int64, and object

# float64: A 64-bit floating-point number. Used for decimal values such as distances,velocities, temperatures, pressures, and any numeric data that includes decimals.

# int64: A 64-bit integer. Used for whole-number values such as number of moons or attributes that happened to convert cleanly to integers.

# object: A generic Python object type. In this DataFrame, object columns contain strings. These are non-numeric columns such as categorical labels or Yes/No fields.
```

```python
semi_major_axis = ((df["Perihelion (10^6 km)"] + df["Aphelion (10^6 km)"]) / 2) * 1e6 * u.km

df["Semi-Major Axis (km)"] = semi_major_axis

print(df.columns)
print(df["Semi-Major Axis (km)"])
```

```python
orbital_cols = [c for c in df.columns if "Orbital Period" in c]
print("Orbital period columns found:", orbital_cols)

if not orbital_cols:
    raise ValueError("No 'Orbital Period' column found in df.columns")

source_col = orbital_cols[0]   

def to_year_quantity(x):
    if hasattr(x, "to"):
        return x.to(u.year)
    else:
        return (x * u.day).to(u.year)

orbital_period_years = df[source_col].apply(to_year_quantity)

df[source_col] = orbital_period_years
if source_col != "Orbital Period (years)":
    df.rename(columns={source_col: "Orbital Period (years)"}, inplace=True)

print(df.columns)

print(df["Orbital Period (years)"])

favorite_planet = "Earth"   # change if you like
value = df.loc[df["Planet"] == favorite_planet, "Orbital Period (years)"].iloc[0]
print(f"\n{favorite_planet}'s orbital period: {value.value:.4f} years")
```
