```python
matrix = np.array([
    [5, 12, 17, 9],
    [8, 21, 14, 3],
    [6, 13, 19, 11]
])
```

```python
print("Shape:", matrix.shape)
```

```python
rows, cols = matrix.shape
print("Rows:", rows)
print("Columns:", cols)
```

```python
print("Third column:", matrix[:, 2])
```

```python
print("Second row:", matrix[1])
```

```python
print(matrix * 2)
```

```python
print(np.mean(matrix[:,0])) # col 1
```

```python
print(np.mean(matrix[:,1])) # col 2
```

```python
print(np.mean(matrix[:,2])) # col 3
```

```python
print(np.mean(matrix[:,3])) # col 4
```

```python
print(np.mean(matrix)) # overall mean
```

```python
print("Max of each row:", np.max(matrix, axis=1))
print("Min of each column:", np.min(matrix, axis=0))
```

```python
flat = matrix.flatten()
print("Flattened array:", flat)
print("New shape:", flat.shape)
```

```python
flat_min_index = np.argmin(flat)
flat_max_index = np.argmax(flat)
print("Flat index of smallest value:", flat_min_index)
print("Flat index of largest value:", flat_max_index)
```

```python
min_coords = np.unravel_index(flat_min_index, matrix.shape)
max_coords = np.unravel_index(flat_max_index, matrix.shape)
print("Coordinates of smallest value:", min_coords)
print("Coordinates of largest value:", max_coords)
```

```python
greater_than_15 = matrix[matrix > 15]
print("Values greater than 15:", greater_than_15)
```

```python
sorted_rows = np.sort(matrix, axis=1)

print(sorted_rows)
```

```python
def zeroArray():
    return np.zeros((8, 8))
```

```python
zeroArray()
```

```python
def oddRowPattern():
    arr = np.zeros((8, 8), dtype=int)
    
    pattern = np.array([1, 0] * 4) 
    
    arr[1] = pattern
    arr[3] = pattern
    arr[5] = pattern
    arr[7] = pattern
    
    return arr
```

```python
oddRowPattern()
```

```python
def checkerboard():
    board = np.zeros((8, 8), dtype=int)
    pattern1 = np.array([1, 0] * 4)
    pattern2 = np.array([0, 1] * 4)
    
    board[0] = pattern1
    board[1] = pattern2
    board[2] = pattern1
    board[3] = pattern2
    board[4] = pattern1
    board[5] = pattern2
    board[6] = pattern1
    board[7] = pattern2
    
    return board
```

```python
checkerboard()
```

```python
def reverse_checkerboard():
    board = np.zeros((8, 8), dtype=int)
    pattern1 = np.array([0, 1] * 4)
    pattern2 = np.array([1, 0] * 4)

    board[0] = pattern1
    board[1] = pattern2
    board[2] = pattern1
    board[3] = pattern2
    board[4] = pattern1
    board[5] = pattern2
    board[6] = pattern1
    board[7] = pattern2

    return board
```

```python
reverse_checkerboard()
```

```python
def expansion(arr, n):
    result = []
    spaces = " " * n     
    
    for word in arr:
        new_word = ""    
        for i in range(len(word)):
            new_word = new_word + word[i]  
            if i != len(word) - 1:
                new_word = new_word + spaces   
        result.append(new_word)
    return np.array(result)
```

```python
expansion("word", 1)
```

```python
def secondDimmest(arr):
    result = []
    
    for row in arr:   
        sorted_row = np.sort(row)
        second_brightest = sorted_row[-2]
        result.append(second_brightest)
    return np.array(result)
```
