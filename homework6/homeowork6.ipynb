```python
def create_checkerboard(array):
    arr = array.copy()
    for i in range(7):
        if i % 2 == 0:
            arr[i,::2] = 1
        else:
            arr[i,1::2] = 1
    return arr
```

```python
new_matrix = create_checkerboard(matrix)
```

```python
print(new_matrix)
```

```python
def create_off_checkerboard(array):
    arr = array.copy()
    for i in range(7):
        if i % 2 == 0:
            arr[i,1::2] = 1
        else:
            arr[i,::2] = 1
    return arr
```

```python
new_matrix1 = create_off_checkerboard(matrix)
```

```python
print(new_matrix1)
```
