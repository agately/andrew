```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.loadtxt("/Users/droogately/python_decal_fall2025/course_assignments/homework7/star_data.txt",
skiprows=1)
```

```python
print(data.shape)
```

```python
magnitude = data[:, 0]
```

```python
temperature = data[:, 1]
```

```python
print("Minimum magnitude = ", np.min(magnitude))
print("Maximum magnitude = ", np.max(magnitude))
```

```python
print("Minimum temperature = ", np.min(temperature))
print("Maximum temperature = ", np.max(temperature))
```

```python
min_temp_index = np.argmin(temperature)
max_temp_index = np.argmax(temperature)
print(data[min_temp_index]) # Coldest star
print(data[max_temp_index]) # Hottest star
```

```python
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)
```

```python
plt.figure(figsize=(8, 6))

# magnitude: smaller = brighter, larger = dimmer
bright_idx = np.argmin(magnitude)
dim_idx    = np.argmax(magnitude)

# temperature: smaller = colder, larger = hotter
cold_idx   = np.argmin(temperature)
hot_idx    = np.argmax(temperature)
# All stars

plt.scatter(log_temperature, log_magnitude, s=10, label="Stars")

# Highlight special stars
plt.scatter(log_temperature[bright_idx], log_magnitude[bright_idx],
            color="black", marker="*", s=300, label="Brightest star")
plt.scatter(log_temperature[dim_idx], log_magnitude[dim_idx],
            color="purple", marker="*", s=300, label="Dimmest star")
plt.scatter(log_temperature[hot_idx], log_magnitude[hot_idx],
            color="green", marker="*", s=300, label="Hottest star")
plt.scatter(log_temperature[cold_idx], log_magnitude[cold_idx],
            color="orange", marker="*", s=300, label="Coldest star")

# Invert axes (HR diagram convention)
ax = plt.gca()
ax.invert_xaxis()  # hotter (larger T, larger log T) toward the left
ax.invert_yaxis()  # brighter (smaller magnitude) toward the top

# Labels, title, legend
plt.title("Hertzsprung–Russell Diagram")
plt.xlabel("Temperature (log K)")
plt.ylabel("Magnitude (log)")
plt.legend()

plt.show()
```
