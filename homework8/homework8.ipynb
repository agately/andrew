```python
median_lat = df["latitude"].median()
median_lon = df["longitude"].median()

print("Median latitude:", median_lat)
print("Median longitude:", median_lon)
```

```python
median_lat = df["latitude"].median()

df["distance1"] = abs(df["latitude"] - median_lat)

print(df.head())
```

```python
median_lon = df["longitude"].median()

df["distance2"] = abs(df["longitude"] - median_lon)

print(df.head())
```

```python
print(df.shape)
print("There are 52 Rows and 6 columns now")
```

```python
# --- Closest to median latitude ---
closest_lat_idx = df["distance1"].idxmin()
closest_lat_state = df.loc[closest_lat_idx, "name"]
closest_lat_value = df.loc[closest_lat_idx, "distance1"]

print(f"State closest to the median latitude: {closest_lat_state} ({closest_lat_value:.2f})")


# --- Farthest from median latitude ---
farthest_lat_idx = df["distance1"].idxmax()
farthest_lat_state = df.loc[farthest_lat_idx, "name"]
farthest_lat_value = df.loc[farthest_lat_idx, "distance1"]

print(f"State farthest from the median latitude: {farthest_lat_state} ({farthest_lat_value:.2f})")


# --- Closest to median longitude ---
closest_lon_idx = df["distance2"].idxmin()
closest_lon_state = df.loc[closest_lon_idx, "name"]
closest_lon_value = df.loc[closest_lon_idx, "distance2"]

print(f"State closest to the median longitude: {closest_lon_state} ({closest_lon_value:.2f})")


# --- Farthest from median longitude ---
farthest_lon_idx = df["distance2"].idxmax()
farthest_lon_state = df.loc[farthest_lon_idx, "name"]
farthest_lon_value = df.loc[farthest_lon_idx, "distance2"]

print(f"State farthest from the median longitude: {farthest_lon_state} ({farthest_lon_value:.2f})")
```

```python


cols = df.columns

x1, y1 = cols[0], cols[1]
x2, y2 = cols[2], cols[3]
hist_col = cols[0]
box_col  = cols[1]

fig, axes = plt.subplots(2, 2, figsize=(10, 8))

sns.scatterplot(data=df, x=x1, y=y1, ax=axes[0, 0])
axes[0, 0].set_title(f"Scatter: {x1} vs {y1}")


sns.scatterplot(data=df, x=x2, y=y2, ax=axes[0, 1])
axes[0, 1].set_title(f"Scatter: {x2} vs {y2}")

sns.histplot(data=df, x=hist_col, ax=axes[1, 0])
axes[1, 0].set_title(f"Histogram of {hist_col}")

sns.boxplot(data=df, x=box_col, ax=axes[1, 1])
axes[1, 1].set_title(f"Boxplot of {box_col}")

plt.show()

#note for later, fix the formatting of the text
```
